```python
import torch
from torch.distributions import Normal, Categorical
```

```python

# 1. Continuous Distribution (Normal)
m = Normal(torch.tensor([0.0]), torch.tensor([1.0])) # Mean 0, Std 1
sample = m.sample()
lp = m.log_prob(sample) 
print(f"Log prob of {sample.item()}: {lp.item()}")

# 2. Discrete Distribution (Categorical)
probs = torch.tensor([0.1, 0.2, 0.7])
m_cat = Categorical(probs)
action = torch.tensor(2) # Choosing the third index
lp_cat = m_cat.log_prob(action)
print(f"Log prob of action 2: {lp_cat.item()}")
```

### Action Example

```python
from torch.nn import functional as F

action_logits = torch.rand(5)
action_probs = F.softmax(action_logits, dim=-1)
action_probs        # 1x5 tensor of probabilities
```

```python
dist = Categorical(action_probs)
action = dist.sample()

print(f"Sampled action: {action.item()}")
print(f"Probability of sampled action: {action_probs[action].item()}")
print(f"Log probability of sampled action: {dist.log_prob(action).item()}")
print(f"Log probability from action_probs: {torch.log(action_probs[action]).item()}")
```

```python
import numpy as np

np.log(action_probs[action].item())
```

Batch example:

```python
action_logits = torch.rand(3,5)
action_probs = F.softmax(action_logits, dim=-1)
action_probs        # 3x5 tensor of probabilities
```

```python
dist = Categorical(action_probs)
action = dist.sample()

print(f"Sampled action: {action}")
```

```python
batch_idx = torch.arange(action_probs.size(0))
print(f"Batch indices: {batch_idx}")

sampled_action_probs = action_probs[batch_idx, action]
sampled_log_probs = torch.log(sampled_action_probs)   # same as dist.log_prob(action)

print("Sampled actions:", action)
print("Probability of sampled actions:", sampled_action_probs)
print("Log probability of sampled actions:", sampled_log_probs)
```
